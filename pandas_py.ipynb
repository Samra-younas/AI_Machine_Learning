# json files
✔ Hierarchical (nested) format, more flexible than CSV.
✔ Stores data as key-value pairs, like a Python dictionary.
✔ Supports nested data, while CSV does not.
✔ Commonly used in APIs, web development, and NoSQL databases.

```python
import pandas as pd
df = pd.read_json('data.json')  # Read a JSON file
print(df)
```

# ✔ CSV is best for simple, structured data (like databases,excel file, reports, tables).
# ✔ JSON is best for web data, APIs, and complex/nested structures.
# ✔ Pandas can handle both CSV and JSON easily! 🚀

```python
import pandas as pd

df = pd.read_json('data.json')

print(df.to_string())
```

# cleaning data

# Note: By default, the dropna() method returns a new DataFrame, and will not change the original.

# put values in place of empty

```python
import pandas as pd

df = pd.read_csv('data.csv')

df.fillna(130, inplace = True)
print(df)
```

# replace only for specified column

```python
import pandas as pd

df = pd.read_csv('data.csv')
print(df.columns)
df["Model"].fillna(130, inplace = True)
print(df.to_string())
```

# correct wrong or misspeled data

```python
df = pd.read_csv('data.csv')

df.loc[7,'Duration'] = 45

print(df.to_string())
```

# removing rows

```python
for x in df.index:
  if df.loc[x, "Duration"] > 120:
    df.drop(x, inplace = True)
```

```python
df.drop_duplicates(inplace = True)
```
